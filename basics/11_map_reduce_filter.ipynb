# Map, Reduce and Filter

- map(function, iterable object)
- filter(function, data)
- reduce(function, data)

## 1. MAP()

### Area of a circle

```python
import math
```

```python
def area(r):
    return math.pi * (r**2)

radii = [1,2,3,4,5]

all_areas = []

for r in radii:
    x = area(r)
    all_areas.append(x)
print(all_areas)
```

```python
# Using map
all_areas = list(map(area, radii))
print(all_areas)
```

### Convert Celsius scale to Fahrenheit

- F = (9/5) * C +32

```python
temps = [("Chicago", 25), ("Dekalb", 18), ("Athens", 20)]

to_F = lambda temp: (temp[0], (9/5) * temp[1] + 32)

list(map(to_F, temps))
```

## 2. FILTER

```python
import statistics

data = [2, 9, 10, 33, 100]

avg = statistics.mean(data)
print(avg)
```

```python
# filter numbers> average

filtered = list(filter(lambda x: x > avg, data))
print(filtered)
```

```python
fruits = ["banana", "", 0, 3, 0, False, True, "apple", "strawberry"]

# filter out null values

print(list(filter(None, fruits)))
```

## 3. REDUCE()

- Returns a single value

```python
from functools import reduce
```

```python
# Multiply all the items in a list

numbers = [3,5,10]

multiplier = lambda x, y: x * y

print(reduce(multiplier, numbers)) # (3*5*10)
```

```python
a = ["Geeks", "for", "Geeks"]
r = reduce(lambda x, y: x + " " + y, a)
print(r)
```

```python
a = [2, 4, 6, 8]
r = reduce(lambda x, y: x + y, a)
print(r)
```
