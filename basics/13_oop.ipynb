# Object Oriented Programming (OOP) - Basics

## 1. Creating Classes and Objects
- Length and breadth as attributes

- init() - constructor of class

- self parameter - refers to the newly created instance of a class


```python
class Rectangle:
    def __init__(self):
        self.length = 10
        self.breadth = 5
```

```python
rect = Rectangle()
rect.breadth
```

### Parameterized constructor

```python
class Rectangle:
    def __init__(self, length, breadth): # instance variable
        self.length = length
        self.breadth = breadth
```

```python
rect = Rectangle(5, 10)
rect.breadth
```

## 2. Class Variable and Instance Variables

```python
class Circle:
    pi = 3.14 # class variable
    def __init__(self, radius):
        self.radius = radius 
```

```python
circle_1 = Circle(5)
circle_1.radius
```

```python
circle_1.pi
```

### How to change the class variable

```python
Circle.pi = 3.1436
```

```python
Circle_1 = Circle.pi
Circle_1 
```

## 3. Adding a Method to a class

- Create a class that calculate the area of an object

```python
class Rectangle:
    def __init__(self, length, breadth): 
        self.length = length
        self.breadth = breadth
    def calculate_area(self):
        return self.length * self.breadth
```

```python
# Instantiating the Rectangle class
rect = Rectangle(10,5)
```

```python
rect.length
```

```python
rect.breadth
```

```python
rect.calculate_area()
```

```python
new_rect = Rectangle(15, 8)
new_rect.calculate_area()
```

## 4. Class Inheritance and Overriding

```python
class Employee:
    def function1(self):
        print("Hello World")
class Department(Employee): #inheriting from the parent class
    pass
```

```python
emp = Employee()
emp.function1()
```

```python
dept = Department()
dept.function1()
```
